## One report for several cells

Start a new session, execute the next two cells, then stop. The full report appears once at the end.

```python
from linescope import profile

session = profile.start(backend="trace", display="end")
```

```python
values = list(range(20_000))
squares = [value * value for value in values]
```

```python
average = sum(squares) / len(squares)
```

```python
profile.stop()
```

```python
session.save("notebook.html")
```
