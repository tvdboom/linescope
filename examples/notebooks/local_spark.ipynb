# LineScope with local Spark

Requires Java and the Spark extra. local[2] starts Spark on this machine without an external cluster.

```python
from pyspark.sql import SparkSession

from linescope import profile

spark = (
    SparkSession.builder.master("local[2]")
    .appName("LineScope notebook")
    .config("spark.sql.shuffle.partitions", "2")
    .getOrCreate()
)
```

```python
with profile(backend="trace", spark=True, display="end") as session:
    values = spark.range(10_000).filter("id % 2 = 0")
    totals = values.groupBy().sum("id").collect()
```

Only the explicit collect triggers the workload. Driver elapsed time and executor/task counters are distinct.
